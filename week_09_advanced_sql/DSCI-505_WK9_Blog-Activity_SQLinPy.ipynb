# Student Practice Tasks 

```python
import sqlite3                        
import pandas as pd                  
conn = sqlite3.connect('my_data.db')  
```

1. SELECT: *Sales Data*

```python
# TASK: Display sale_date, product, and price from the sales table.

# Read sales.csv into the sales table
pd.read_csv("sales.csv").to_sql('sales', conn, if_exists='replace', index=False)

# Select the three columns
pd.read_sql_query('''
SELECT sale_date, product, price FROM sales
''', conn)
```

2. WHERE: *Human Resources*

```python
# TASK: Find employees in Chicago with more than 10 years of experience using WHERE and AND.

# Read hr_employees.csv into the 'employees' table
pd.read_csv('hr_employees.csv').to_sql('employees', conn, if_exists='replace', index=False)
pd.read_sql_query('''
SELECT name, department, city, years_experience
FROM employees
WHERE years_experience > 10 AND city = 'Chicago'
''', conn)
```

3. ORDER BY: *Real Estate*

```python
# TASK: Display the three largest properties by square footage using ORDER BY and LIMIT.

# Read realestate_listings.csv into 'listings' table
pd.read_csv('realestate_listings.csv').to_sql('listings', conn, if_exists='replace', index=False)
pd.read_sql_query('''
SELECT *
FROM listings
ORDER BY sqft DESC
LIMIT 3
''', conn)
```

4. GROUP BY: *Marketing*

```python
# TASK: Calculate total weekly impressions using SUM() and GROUP BY.

# Read marketing_campaigns.csv into 'campaigns' table
pd.read_csv('marketing_campaigns.csv').to_sql('campaigns', conn, if_exists='replace', index=False)
pd.read_sql_query('''
SELECT week, SUM(impressions) AS total_impressions
FROM campaigns
GROUP BY week
''', conn)
```

5. HAVING: *Restaurant*

```python
# TASK: Find dates when total restaurant sales exceeded $400.

# Read restaurant_orders.csv into 'orders' table
pd.read_csv('restaurant_orders.csv').to_sql('orders', conn, if_exists='replace', index=False)
pd.read_sql_query('''
SELECT order_date, SUM(order_total) AS day_sales
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
''', conn)
```

6. Aggregate Functions: *Healthcare*

```python
# TASK: Find the average patient age and longest waiting time for each doctor.

# Read clinic_visist.csv into 'visits' table
pd.read_csv('clinic_visits.csv').to_sql('visits', conn, if_exists='replace', index=False)
pd.read_sql_query('''
SELECT doctor, COUNT(*) AS total_visits,
       ROUND(AVG(wait_minutes), 1) AS avg_wait_minutes
FROM visits
GROUP BY doctor
''', conn)
       
```

7. JOIN: *Banking*

```python
# TASK: Join customer and transaction tables to calculate total transaction amounts by branch.

# Read bank_transactions.csv and bank_customers.csv each to their own tables - 'transactions' and 'customers', respectively
pd.read_csv('bank_transactions.csv').to_sql('transactions', conn, if_exists='replace', index=False)
pd.read_csv('bank_customers.csv').to_sql('customers', conn, if_exists='replace', index=False)
pd.read_sql_query('''
SELECT c.branch, SUM(amount) AS total_transactions
FROM transactions AS t
JOIN customers AS c ON t.customer_id = c.customer_id
GROUP BY c.branch
''', conn)
```
